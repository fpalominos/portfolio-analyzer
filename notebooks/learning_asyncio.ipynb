Learning Asyncio

```python
import asyncio
from datetime import datetime

async def hello():
    print("Starting hello")
    await asyncio.sleep(2)
    print("Finished hello")

await hello()
```

```python
async def fetch_price(symbol: str):
    print(f"fetching {symbol}")
    await asyncio.sleep(2)
    print(f"Received {symbol}")

start = datetime.now()
print(f"start: {start.isoformat()}")

await asyncio.gather(
    fetch_price("a"),
    fetch_price("b"),
    fetch_price("c"),
    fetch_price("d"),
    fetch_price("e"),
    fetch_price("f")
)
end = datetime.now()
print(f"end: {end.isoformat()}")

elapsed = end - start
print(f"elapsed: {elapsed} ({elapsed.total_seconds():.3f}s)")
```

```python
import asyncio
import httpx
"""
>>> params = {'key1': 'value1', 'key2': 'value2'}
>>> r = httpx.get('https://httpbin.org/get', params=params)

https://www.python-httpx.org/quickstart/
"""
params = {'function': 'REALTIME_BULK_QUOTES', 'symbol': 'MSFT,AAPL,IBM', 'apikey': 'EJ3DAN05AD7BROPU'}

request = "https://www.alphavantage.co/query?"
# request = "https://www.alphavantage.co/query?function=REALTIME_BULK_QUOTES&symbol=MSFT,AAPL,IBM&apikey=EJ3DAN05AD7BROPU"
async with httpx.AsyncClient() as client:
    #response = await client.get(request)
    response = await client.get(request, params=params)

    print(response.json())
```
