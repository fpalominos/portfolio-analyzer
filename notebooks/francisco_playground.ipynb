```python
numbers = [1, 2, 3, 4, 5]

# Use for at the end (generator/list comprehension) when you're:
# 1. Passing to a function (most common):
sum(n for n in numbers)
max(n for n in numbers)
any(n > 5 for n in numbers)

# 2. Creating a new list/set/dict — use [], {}, etc:
squared = [n ** 2 for n in numbers]
print(squared)

# 3. One-liner transformations
strings = ["abd", "de"]
result = [x.upper() for x in strings]
filtered = [x for x in numbers if x > 5]


#Use traditional for loop when:
#1.Complex multi-line logic:
# for n in numbers:
#     print(f"Processing {n}")
#     result = expensive_calculation(n)
#     store_result(result)
# 2.Multiple statements per iteration:
# for user in users:
#     validate(user)
#     save(user)
#     notify(user)
# 3.Early exit needed:
# for item in items:
#     if found(item):
#         break

def unnecessary_sum(numbers: list[
    int]) -> int:  # for learning purposes only. A generator must be used instead i.e. sum(n for n in numbers)
    acc = 0
    for n in numbers:
        acc += n
    return acc


print(f"unnecessary_sum: {unnecessary_sum(numbers)}")
```

```python
# zip is lazy, see zip(portfolio.positions, prices)
# zip is lazy (returns an iterator). It only produces pairs when consumed.
# To realize (execute) it you can:
# materialize to a list: values = list(zip(...))
# materialize to a tuple: values = tuple(zip(...))
# iterate over it directly: for pos, price in zip(...): ...
# pass it to a consuming function: max(zip(...), key=...), sum(...), list(...), any(...), etc.
# use unpacking / comprehension: pairs = [*zip(...)] or pairs = [(p,s) for p,s in zip(...)]
```

```python
import asyncio

from portfolio_analyzer.services.price_service import PriceService
from portfolio_analyzer.domain.portfolio import Portfolio
from portfolio_analyzer.domain.portfolio import Stock

dict = ["AAPL", "AAPL", "NVDA"]

price_service = PriceService()


async def largest_position(price_service: PriceService, portfolio: Portfolio):
    tasks = [
        price_service.get_price(stock.symbol)
        for stock in portfolio.positions
    ]

    prices = await asyncio.gather(*tasks)

    values_by_position: list[tuple[str, Any]] = [(stock.symbol, (stock.shares * price)) for stock, price in
                                                 zip(portfolio.positions, prices)]

    return max(
        values_by_position,
        key=lambda p: p[1],
        default=None
    )


portfolio = (
    Portfolio()
    .add_position(Stock("AAPL", 10))
    .add_position(Stock("MSFT", 5))
    .add_position(Stock("NVDA", 3))
)

price_service = PriceService()
await largest_position(price_service, portfolio)

```

```python
data = {"error": "Rate limit exceeded", "k2": "value2"}
error_exists = "error" in data  # sames as data.__contains__("error") but more pythonic

print(error_exists)


class InvalidApikeyError(Exception):
    pass


class RateLimitExceededError(Exception):
    pass


class PriceServiceError(Exception):
    pass

if "error" in data:

    match data["error"]:

        case "API key invalid":
            raise InvalidApikeyError()

        case "Rate limit exceeded":
            raise RateLimitExceededError()

        case _:
            raise PriceServiceError(data["error"])
```
